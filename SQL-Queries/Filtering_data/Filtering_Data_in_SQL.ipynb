# **Using upper() and lower() functions**

```python
query = """
SELECT *,CONCAT(upper(customer_first_name)," ",lower(customer_last_name)) as customer_new_name
FROM `scaler-sql-476312.farmers_market.customer`
"""

df = client.query(query).to_dataframe()
df.head()
```

# **Using Substring function**

```python
query = """
SELECT customer_first_name,customer_last_name,CONCAT(substr(customer_first_name,1,1),".",customer_last_name) as customer_full_name
FROM `scaler-sql-476312.farmers_market.customer`
"""

df = client.query(query).to_dataframe()
df.head()
```

```python
query = """
SELECT customer_first_name,customer_last_name,CONCAT(lower(substr(customer_first_name,1,1)),upper(substr(customer_first_name,2))," ",lower(substr(customer_last_name,1,1)),
upper(substr(customer_last_name,2))) as name
FROM `scaler-sql-476312.farmers_market.customer`
"""

df = client.query(query).to_dataframe()
df.head()
```

# Trim Function


Basically it will remove all the blanks and it will act as a null but it is not a null

For Suppose we have a 5 blanks by using trim function it will remove the all the 5 blanks and length will be zero after that it will be  a empty which means no spaces just empty string

```python
#Question -- From Above Question we will remove the blanks for that product by using trim function
query = """
SELECT * FROM `farmers_market.product`
WHERE trim(product_size) = "" or product_size is null
"""
df = client.query(query).to_dataframe()
df
```

```python
#Question -- Replacing null values with unknown and created a cleaned_ps column to store it by using ifnull() function
query = """
SELECT
  product_id,
  product_name,
  product_size, ifnull(product_size, 'unknown') AS cleaned_ps
FROM `farmers_market.product`
"""
df = client.query(query).to_dataframe()
df
```
